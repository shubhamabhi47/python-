```python
def most_freq(s):
    frequency = {}
    
    # Count the frequency of each character
    for char in s:
        if char in frequency:
            frequency[char] += 1
        else:
            frequency[char] = 1
    
    # Find the most frequent character
    max_count = 0
    most_frequent = None
    for char, count in frequency.items():
        if count > max_count:
            max_count = count
            most_frequent = char
    
    return most_frequent
```

```python
def most_frequent_char(s):
    # Create a dictionary to count the frequency of each character
    freq = {}
    for char in s:
        if char in freq:
            freq[char] += 1
        else:
            freq[char] = 1
    
    # Find the character with the maximum frequency
    most_frequent = max(freq, key=freq.get)
    
    return most_frequent
```

```python
import string

def most_frequent(word):
    # Convert the word to lowercase to handle case sensitivity
    word = word.lower()
    
    # Get the alphabet and initialize a dictionary to count frequency
    alphabet = string.ascii_lowercase
    dictionary = {}

    # Initialize all alphabet letters to zero
    for letter in alphabet:
        dictionary[letter] = 0

    # Count each letter in the word, ignoring non-alphabetic characters
    for letter in word:
        if letter in dictionary:
            dictionary[letter] += 1

    # Sort dictionary items by frequency (high to low) and alphabetically in case of ties
    dictionary = sorted(dictionary.items(), reverse=True, key=lambda x: (x[1], x[0]))

    # Print the most frequent letters until the frequency changes
    for position in range(len(dictionary)):
        print(dictionary[position])
        if position != len(dictionary) - 1:
            if dictionary[position + 1][1] < dictionary[position][1]:
                break

# Taking user input
user_string = input("Enter any string to find most frequent character: ")
most_frequent(user_string)
```
