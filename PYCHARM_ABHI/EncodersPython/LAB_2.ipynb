```python
# append(), insert(position, value), extend()
# l2 = [6,7,9,4,3]
l1 = [1,2,3,4,5]
l1.extend([1 ,2 ,3])
print(l1)
# l1.extend("123")
l1.extend("1 2 3")
print(l1)
```

```python
# to delete an element from list --> remove(), pop()
list = [3,4,5,6,7]
print(list)
list.remove(3) #it will take element
print(list)
list.pop(3)  #it will take index
print(list)
```

```python
# for printing substring of any string
list = [1,2,3,4,5,6,7,8]
print(list[:])
print(list[1:3])
print(list[:3])
print(list[:-4])
print(list[::3])
print(list[::-1])
```
