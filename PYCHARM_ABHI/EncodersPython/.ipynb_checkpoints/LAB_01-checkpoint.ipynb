```python
# single or double quotes within a string  can be escaped using backslash character
stats="Hello! \"I am Shubham.\" How was your day?"
```

```python
# multiline string --> string in enclosed in triple-single or double quotes
stats = """Hyy!
I am Shubham. 
How was your day?"""
print(stats)
```

```python
# String formating operator
age = 20
print('My age is %d ' %(age))
```

```python
# String slicing 
R = 'I am Shubham'
print(R[0 : 7 : 2])
print(R[7:1:-2])
print(R[::-1])
print(R[::])
```

```python
s = 'abcabcab'
print(s[ 1::3])
print(s[ :5:3])
print(s[ ::-1])
# s='abc'*3
```

```python
# Interpolation is the insertion of something of different nature into something else
# The process of inserting an object into a predefined string is called string interpolation or string formatting
# Oldest technique
name = 'Shubham'
'My name is %s.' % name
```

```python
# %-Formatting
name = 'Shubham'
city = 'Patna'
'My name is %s. and I live in %s.' %(name , city)
```

```python
# str.format()
name = 'Shubham'
city = 'Patna'
print('My name is {}. and I live in {}.'. format(name , city))
print('My name is {0}. and I live in {1}.'. format(name , city))
```

```python
n = "Shubham"
c = "Patna"
print('My name is {name}. and I live in {city}.'. format(name=n , city=c))
```
