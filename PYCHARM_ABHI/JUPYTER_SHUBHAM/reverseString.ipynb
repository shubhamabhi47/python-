```python
my_string = "Hello, World!"
reversed_string = my_string[::-1]
print(reversed_string)
```

```python
my_string = "Hello, World!"
reversed_string = ""
for char in my_string:
    reversed_string = char + reversed_string
print(reversed_string)
```

```python
my_string = "Hello, World!"
reversed_string = ''.join(reversed(my_string))
print(reversed_string)
```

```python
def reverse_string(s):
    if len(s) == 0:
        return s
    else:
        return reverse_string(s[1:]) + s[0]

my_string = "Hello, World!"
print(reverse_string(my_string))
```

```python
my_string = "Hello, World!"
stack = list(my_string)  # Convert string to list (acting as a stack)
reversed_string = ''

while stack:
    reversed_string += stack.pop()  # Pop from stack and append to result

print(reversed_string)
```

```python
from functools import reduce

my_string = "Hello, World!"
reversed_string = reduce(lambda x, y: y + x, my_string)
print(reversed_string)
```

```python
my_string = "Hello, World!"
reversed_string = ''.join(my_string[i] for i in range(len(my_string) - 1, -1, -1))
print(reversed_string)
```

```python
my_string = "Hello, World!"
reversed_string = bytearray(my_string, 'utf-8')
reversed_string.reverse()
print(reversed_string.decode('utf-8'))
```

```python
def reverse(s):
    n = len(s)
    reversed_s = [''] * n
    for i in range(n):
        reversed_s[n - 1 - i] = s[i]
    return ''.join(reversed_s)

my_string = "Hello, World!"
print(reverse(my_string))
```
