```python
def most_freq (s):
    frequency = {}
    for char in s:
        if char in frequency:
            frequency[char] += 1
        else:
            frequency[char] = 1
   
    max_count = 0
    most_frequent = None
    for char, count in frequency.items():
        if count > max_count:
            max_count = count
            most_frequent = char
```

```python
def is_prime(n):
    if n <= 1:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

# Input from user
p = int(input("Enter a number: "))

if is_prime(p):
    print(f"{p} is a prime number.")
else:
    print(f"{p} is not a prime number.")
```
