```python
w = float(input("Enter your weight in kg: "))
h = float(input("Enter your height in meters: "))


bmi = w / (h ** 2)


if bmi < 18.5:
    print("Underweight")
elif 18.5 <= bmi < 25:
    print("Normal weight")
elif 25 <= bmi < 30:
    print("Slightly overweight")
elif 30 <= bmi < 35:
    print("Obese")
else:
    print("Clinically obese")
```

```python
# wap to take day as a var name and show the respetive name in a way in python
# take a char as a input show whether it is vowel or not

# Program to display the name of the day based on the input number using match-case

# Take the day number as input from the user
day = int(input("Enter the day number : "))

match day:
    case 1:
        print(" Monday")
    case 2:
        print("Tuesday")
    case 3:
        print(" Wednesday")
    case 4:
        print(" Thursday")
    case 5:
        print("Friday")
    case 6:
        print(" Saturday")
    case 7:
        print(" Sunday")
    case _:
        print("Invalid input ")
```

```python
# take any num as input and find sum of digit of number
# Take input from the user and reverse it and check it is same or not in python
num = int(input("Enter a number: "))

sum = 0

while num > 0:
    sum += num % 10  
    num //= 10  

print("Sum of digits:", sum)
```

```python
num = int(input("Enter a number: "))

original_num = num

reversed_num = 0

while num > 0:
    digit = num % 10  
    reversed_num = reversed_num * 10 + digit  
    num //= 10  
print("reversed num is",reversed_num)
if (original_num == reversed_num):
    print(" palindrome")
else:
    print(" not a palindrome")
```

```python
num = int(input("Enter a number: "))

num_str = str(num)
num_len = len(num_str)

sum_of_powers = 0
temp_number = num

while temp_number > 0:
    digit = temp_number % 10  
    sum_of_powers += digit ** num_len  
    temp_number //= 10  

if sum_of_powers == num:
    print("Armstrong number.")
else:
    print(f"{num} is not an Armstrong number.")
```
