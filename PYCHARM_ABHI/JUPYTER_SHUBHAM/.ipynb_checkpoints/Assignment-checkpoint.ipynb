```python
#1. WAP to make a calculator do all arithmetic operations.
a = float(input("Enter first number: "))
b = float(input("Enter second number: "))
# calculator(a, b)
def calculator(a, b):
    print("Addition:", a + b)
    print("Subtraction:", a - b)
    print("Multiplication:", a * b)
    print("Division:", a / b)
    print("Floor: ",a//b)
calculator(a, b)

```

```python
# 2. WAP to show all the works of the bitwise operator.
# Bitwise Operations

x = int(input("Enter first number: "))
y = int(input("Enter second number: "))

print("Bitwise AND: ", x & y)
print("Bitwise OR: ", x | y)
print("Bitwise XOR: ", x ^ y)
print("Bitwise NOT of x: ", ~x)
print("Bitwise NOT of y: ", ~y)
print("Bitwise Left Shift x by 2: ", x << 2)
print("Bitwise Right Shift x by 2: ", x >> 2)
```

```python
# 3. WAP to find the simple interest and compound interest where Principal (P), Rate (R), and Time (T) are given by the user.

# Simple Interest
P = float(input("Enter the principal amount: "))
R = float(input("Enter the rate of interest: "))
T = float(input("Enter the time (in years): "))

simple_interest = (P * R * T) / 100
print("Simple Interest: ", simple_interest)

# Compound Interest
compound_interest = P * ((1 + R / 100) ** T - 1)
print("Compound Interest: ", compound_interest)
```

```python
# 4. WAP to find the area and perimeter of a circle and rectangle.
def rectangle_properties(length, width):
    area = length * width
    perimeter = 2 * (length + width)
    print("Rectangle Area:", area)
    print("Rectangle Perimeter:", perimeter)

length = float(input("Enter the length of the rectangle: "))
width = float(input("Enter the width of the rectangle: "))
rectangle_properties(length, width)
```
