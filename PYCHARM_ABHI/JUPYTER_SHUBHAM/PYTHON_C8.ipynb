```python
numbers = list(range(1, 11))

def is_prime(num):
    if num <= 1:
        return False
    for i in range(2, int(num ** 0.5) + 1):
        if num % i == 0:
            return False
    return True

prime_num = [num for num in numbers if is_prime(num)]

print("Prime numbers from the list are:", prime_num)
```
